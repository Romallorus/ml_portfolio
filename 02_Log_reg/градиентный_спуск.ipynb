```python
import numpy as np
import matplotlib.pyplot as plt
```

```python

x = np.linspace(-10, 10, 100)
def f(x):
    return x ** 2 + x * 3 + 5 * np.sin(x)
plt.plot(x, f(x))
```

```python
def f(x):
    return x ** 2 + x * 3 + 5 * np.sin(x)

def f_grad(x):
    return 2*x + 3 + 5*np.cos(x) 

eps = 1e-6
n = 0.1
x = -10.0
max_iter = 1000
for i in range(max_iter):
    grad = f_grad(x)
    x_new = x - n * grad
    if abs(x_new - x) < eps:
        print(f"Сошлись за {i+1} итераций")
        break
    x = x_new
else:
    print("Достигнуто максимальное число итераций")

print(f"Найденный минимум: x = {x:.6f}, f(x) = {f(x):.6f}")

# Визуализация
x_vals = np.linspace(-10, 10, 200)
y_vals = f(x_vals)
plt.plot(x_vals, y_vals, label='f(x)')
plt.scatter(x, f(x), color='red', zorder=5, label='Найденный минимум')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.legend()
plt.grid(True)
plt.show()
```
