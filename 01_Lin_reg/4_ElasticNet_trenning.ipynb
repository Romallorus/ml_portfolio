```python
from sklearn.datasets import load_diabetes
```

```python
diabetes = load_diabetes()
X = diabetes.data
y = diabetes.target
```

```python
diabetes_df = pd.DataFrame(X, columns=diabetes.feature_names)
diabetes_df['PROGRESSION'] = y
```
