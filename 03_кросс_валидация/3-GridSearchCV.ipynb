# Поиск по сетке

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"Advertising.csv")

df.head()

X = df.drop('sales', axis=1)
y = df['sales']

from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = .3, random_state=101) # можно указать поменьше, обучение все равно на всех данных, которые в трейн

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
from sklearn.linear_model import ElasticNet
```

```python
el_net = ElasticNet()   # ключи как в параметрах модели
param_grid = {
    'alpha': [0.1, 1, 2, 3, 4, 4.5, 5, 50, 100],
    'l1_ratio': [.1, .3, .5, .6, .95, .99, 1],
    'max_iter': [1000000, 1000],
}
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
grid_model = GridSearchCV(estimator=el_net,
                          param_grid=param_grid,
                          scoring='neg_mean_squared_error',
                          cv=5,
                          verbose=3)
```

```python
grid_model.fit(X_train, y_train)
```

```python
grid_model.best_estimator_
```

```python
grid_model.cv_results_
```

```python
pd.DataFrame(grid_model.cv_results_)
```

```python
y_pred = grid_model.predict(X_test)
from sklearn.metrics import mean_squared_error
```

```python
mean_squared_error(y_test, y_pred)
```

```python
grid_model.best_score_
```
